## 7. Vòng lặp `for` và `while`

````python
capital = 100
year = 0

while capital < 200:
    capital *= 1.10
    year += 1

print("So nam:", year)
print("Gia tri:", round(capital, 2))
````

## 8. Hàm – `def`


````python
def future_value(pv, r, n):
    return pv * (1 + r) ** n

print(future_value(100, 0.10, 7))
````

### Tham số mặc định


````python
def compound(pv, r=0.08, n=5):
    return pv * (1 + r) ** n

print(compound(100))
print(compound(100, r=0.10))
print(compound(100, n=10))
````

## 9. List comprehension


````python
returns = [0.05, 0.08, -0.02, 0.10]

positive_returns = [r for r in returns if r > 0]
print(positive_returns)
````

Cú pháp:

```python

```

## 18. Broadcasting


Broadcasting cho phép NumPy thực hiện phép toán giữa các mảng có kích thước khác nhau khi chúng tương thích.

## 19. `axis` – rất quan trọng


````python
x = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Tong tat ca:", x.sum())
print("Tong theo cot (axis=0):", x.sum(axis=0))
print("Tong theo hang (axis=1):", x.sum(axis=1))
````

### Mẹo nhớ

- `axis=0`: **đi xuống các hàng**, trả kết quả cho từng cột.
- `axis=1`: **đi ngang các cột**, trả kết quả cho từng hàng.


## 20. Thống kê với NumPy

# PHẦN III – MATPLOTLIB


Matplotlib là thư viện trực quan hóa cơ bản của Python.

Quy trình thường dùng:

```python
fig, ax = plt.subplots()
ax.plot(...)
ax.set_title(...)
ax.set_xlabel(...)
ax.set_ylabel(...)
plt.show()
```


## 23. Line chart – biểu đồ đường


````python
months = np.arange(1, 7)
revenue = np.array([100, 120, 115, 135, 150, 165])

fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(months, revenue, marker="o")
ax.set_title("Revenue by Month")
ax.set_xlabel("Month")
ax.set_ylabel("Revenue")
ax.grid(True, alpha=0.3)

plt.show()
````

## 24. Vẽ nhiều đường trên cùng biểu đồ


````python
profit = np.array([20, 25, 22, 30, 34, 40])

fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(months, revenue, marker="o", label="Revenue")
ax.plot(months, profit, marker="s", label="Profit")

ax.set_title("Revenue and Profit")
ax.set_xlabel("Month")
ax.legend()
ax.grid(True, alpha=0.3)

plt.show()
````

## 25. Bar chart – biểu đồ cột


````python
categories = ["A", "B", "C", "D"]
sales = [120, 180, 90, 150]

fig, ax = plt.subplots(figsize=(6, 4))

ax.bar(categories, sales)
ax.set_title("Sales by Category")
ax.set_xlabel("Category")
ax.set_ylabel("Sales")

plt.show()
````

## 26. Histogram – biểu đồ phân phối


````python
rng = np.random.default_rng(42)
returns = rng.normal(loc=0.001, scale=0.02, size=500)

fig, ax = plt.subplots(figsize=(7, 4))

ax.hist(returns, bins=30, edgecolor="black")
ax.set_title("Distribution of Daily Returns")
ax.set_xlabel("Return")
ax.set_ylabel("Frequency")

plt.show()
````

Histogram đặc biệt hữu ích để xem:

- phân phối lợi suất,
- doanh thu,
- chi phí,
- giá trị giao dịch.


## 27. Scatter plot – biểu đồ phân tán


````python
risk = np.array([0.10, 0.15, 0.12, 0.25, 0.20])
expected_return = np.array([0.06, 0.09, 0.07, 0.14, 0.11])

fig, ax = plt.subplots(figsize=(6, 4))

ax.scatter(risk, expected_return, s=80)
ax.set_title("Risk vs Expected Return")
ax.set_xlabel("Risk")
ax.set_ylabel("Expected Return")
ax.grid(True, alpha=0.3)

plt.show()
````

## 28. Pie chart – biểu đồ tròn


````python
weights = [40, 30, 20, 10]
labels = ["Stocks", "Bonds", "Cash", "Gold"]

fig, ax = plt.subplots(figsize=(6, 6))

ax.pie(weights, labels=labels, autopct="%1.1f%%", startangle=90)
ax.set_title("Portfolio Allocation")

plt.show()
````

## 41. `groupby()`


````python
sales_df = pd.DataFrame({
    "Region": ["South", "North", "South", "Central", "North", "South"],
    "Revenue": [120, 90, 150, 70, 110, 180],
    "Profit": [25, 18, 35, 12, 22, 45]
})

sales_df
````

````python
sales_df.groupby("Region")["Revenue"].sum()
````

### `agg()` – nhiều chỉ số cùng lúc


````python
summary = (
    sales_df
    .groupby("Region")
    .agg(
        Revenue=("Revenue", "sum"),
        Profit=("Profit", "sum"),
        Transactions=("Revenue", "size")
    )
)

summary
````

## 42. Pivot table


````python
transactions = pd.DataFrame({
    "Region": ["South", "South", "North", "North", "Central", "Central"],
    "Category": ["A", "B", "A", "B", "A", "B"],
    "Revenue": [120, 80, 100, 90, 70, 60]
})

pd.pivot_table(
    transactions,
    values="Revenue",
    index="Region",
    columns="Category",
    aggfunc="sum",
    fill_value=0
)
````

## 43. `merge()` – ghép bảng


````python
customers = pd.DataFrame({
    "CustomerID": [1, 2, 3],
    "CustomerName": ["An", "Binh", "Chi"]
})

orders = pd.DataFrame({
    "OrderID": [101, 102, 103, 104],
    "CustomerID": [1, 1, 2, 3],
    "Revenue": [100, 150, 200, 120]
})

pd.merge(orders, customers, on="CustomerID", how="left")
````

## 45. Dữ liệu thời gian


````python
dates = pd.date_range("2026-01-01", periods=6, freq="ME")

time_df = pd.DataFrame({
    "Date": dates,
    "Revenue": [100, 110, 105, 120, 125, 140]
})

time_df
````

````python
time_df["Year"] = time_df["Date"].dt.year
time_df["Month"] = time_df["Date"].dt.month

time_df
````

## 46. Rolling – trung bình trượt


````python
time_df["MA3"] = time_df["Revenue"].rolling(3).mean()

time_df
````

- lợi suất kỳ vọng,
- độ biến động.


````python
portfolio = pd.DataFrame({
    "Stock": ["AAA", "BBB", "CCC", "DDD", "EEE"],
    "Price": [25_000, 42_000, 18_000, 65_000, 30_000],
    "Quantity": [1000, 500, 1500, 300, 800],
    "ExpectedReturn": [0.10, 0.08, 0.15, 0.06, 0.12],
    "Volatility": [0.20, 0.15, 0.30, 0.10, 0.25],
    "Sector": ["Bank", "Bank", "Tech", "Utility", "Retail"]
})

portfolio
````

## Bước 1. Tính giá trị từng vị thế


````python
portfolio["MarketValue"] = portfolio["Price"] * portfolio["Quantity"]

portfolio[["Stock", "MarketValue"]]
````

## Bước 2. Tính trọng số danh mục


````python
total_value = portfolio["MarketValue"].sum()

portfolio["Weight"] = portfolio["MarketValue"] / total_value

portfolio[["Stock", "MarketValue", "Weight"]]
````

## Bước 3. Lợi suất kỳ vọng của danh mục


````python
portfolio_return = np.sum(
    portfolio["Weight"].to_numpy()
    * portfolio["ExpectedReturn"].to_numpy()
)

print(f"Expected portfolio return: {portfolio_return:.2%}")
````

## Bước 4. Phân tích theo ngành


````python
sector_summary = (
    portfolio
    .groupby("Sector")
    .agg(
        MarketValue=("MarketValue", "sum"),
        MeanExpectedReturn=("ExpectedReturn", "mean"),
        MeanVolatility=("Volatility", "mean")
    )
)

sector_summary
````

## Bước 5. Biểu đồ Risk – Return


````python
fig, ax = plt.subplots(figsize=(7, 5))

for sector, group in portfolio.groupby("Sector"):
    ax.scatter(
        group["Volatility"],
        group["ExpectedReturn"],
        s=group["Weight"] * 2500,
        alpha=0.7,
        label=sector
    )

for _, row in portfolio.iterrows():
    ax.annotate(
        row["Stock"],
        (row["Volatility"], row["ExpectedReturn"]),
        xytext=(5, 5),
        textcoords="offset points"
    )

ax.set_title("Risk – Return of Portfolio")
ax.set_xlabel("Volatility")
ax.set_ylabel("Expected Return")
ax.legend()
ax.grid(True, alpha=0.3)

plt.show()
````

## Bước 6. Biểu đồ trọng số danh mục


````python
fig, ax = plt.subplots(figsize=(7, 4))

ax.bar(portfolio["Stock"], portfolio["Weight"])
ax.set_title("Portfolio Weights")
ax.set_xlabel("Stock")
ax.set_ylabel("Weight")

plt.show()
````
