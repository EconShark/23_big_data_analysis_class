# 5. Ví dụ tài chính: Future Value

Giả sử:

- Giá trị hiện tại: \(PV\)
- Lãi suất mỗi kỳ: \(r\)
- Số kỳ: \(n\)

Giá trị tương lai:

\[
FV = PV(1+r)^n
\]

Ví dụ đầu tư 100 đơn vị, lãi suất 10%/năm trong 7 năm.


````python
pv = 100
r = 0.10
n = 7

fv = pv * (1 + r) ** n

print("Future Value =", round(fv, 2))
````

### Giải thích

`(1 + r) ** n` là hệ số tăng trưởng kép sau `n` năm.

Với lãi suất 10%:

- Cuối năm 1: `100 × 1.1`
- Cuối năm 2: `100 × 1.1²`
- ...
- Cuối năm 7: `100 × 1.1⁷`


### Bài tập 1

Một nhà đầu tư có **200 triệu đồng**, kỳ vọng lợi suất **8%/năm**.

Hãy tính giá trị khoản đầu tư sau **5 năm**.


````python
pv = 200_000_000
r = 0.08
n = 5

fv = pv * (1 + r) ** n

print(f"Giá trị tương lai sau {n} năm: {fv:,.0f} đồng")
````

**Nhận xét:** Với lãi suất 8%/năm và lãi kép, giá trị khoản đầu tư tăng theo hàm mũ chứ không tăng tuyến tính.

## 6.4. Dictionary

Dictionary lưu dữ liệu theo dạng:

```text
key → value
```

Ví dụ:

```text
mã tài khoản → thông tin tài khoản
```


````python
accounts = {
    "1001": {
        "name": "Cash",
        "balance": 500.00,
        "type": "Asset"
    },
    "1002": {
        "name": "Accounts Receivable",
        "balance": 1000.00,
        "type": "Asset"
    },
    "2001": {
        "name": "Accounts Payable",
        "balance": 750.00,
        "type": "Liability"
    }
}

for account_number, account_info in accounts.items():
    print("Account number:", account_number)
    print("Account name:", account_info["name"])
    print("Balance:", account_info["balance"])
    print("Type:", account_info["type"])
    print("-" * 30)
````

Cấu trúc điều kiện cho phép chương trình thực hiện hành động khác nhau tùy vào dữ liệu.

### Ví dụ: cần bao lâu để vốn tăng gấp đôi?


````python
capital = 100
rate = 0.10

value = capital
year = 0

while value < capital * 2:
    value = value * (1 + rate)
    year += 1

print("Số năm để vốn ít nhất gấp đôi:", year)
print("Giá trị cuối:", round(value, 2))
````

## 9.1. Hàm tính Future Value


````python
def get_future_value(pv, r, n):
    fv = pv * (1 + r) ** n
    return fv


fv = get_future_value(100, 0.10, 7)

print(round(fv, 2))
````

### Ý nghĩa

Khi đã viết hàm, ta có thể tính cho nhiều khoản đầu tư khác nhau mà không cần viết lại công thức.


````python
investments = [
    (100, 0.10, 7),
    (200, 0.08, 5),
    (500, 0.06, 10)
]

for pv, r, n in investments:
    fv = get_future_value(pv, r, n)
    print(f"PV={pv}, r={r:.0%}, n={n} → FV={fv:.2f}")
````

# 10. Lớp và đối tượng

Python là ngôn ngữ hỗ trợ lập trình hướng đối tượng.

- **Class**: bản thiết kế
- **Object**: một đối tượng cụ thể tạo ra từ class

Ví dụ class `Stock` có thể mô tả một cổ phiếu.


````python
class Stock:
    def __init__(self, symbol, price, quantity):
        self.symbol = symbol
        self.price = price
        self.quantity = quantity

    def market_value(self):
        return self.price * self.quantity


stock1 = Stock("ABC", 25_000, 1000)

print("Mã:", stock1.symbol)
print("Giá:", stock1.price)
print("Số lượng:", stock1.quantity)
print("Giá trị thị trường:", stock1.market_value())
````

### Giải thích

`__init__()` được gọi khi tạo đối tượng.

`self` đại diện cho chính đối tượng đang làm việc.

# 14. Xử lý ngoại lệ với `try...except`

Trong thực tế chương trình có thể gặp lỗi:

- Người dùng nhập sai
- File không tồn tại
- Dữ liệu không chuyển được sang số
- Chia cho 0

`try...except` giúp chương trình xử lý lỗi thay vì dừng đột ngột.


````python
try:
    x = float("abc")
except Exception as e:
    print("Có lỗi:", e)
````

## 14.1. `try...except...else...finally`

- `try`: code có khả năng phát sinh lỗi
- `except`: chạy khi có lỗi
- `else`: chạy khi không có lỗi
- `finally`: luôn chạy


````python
try:
    price = float("10")
    shares = int("2")
    total = price * shares

except Exception as e:
    print("Exception error:", str(e))

else:
    print("The total value of the shares is:", total)

finally:
    print("Finished.")
````

# 15. Bài tập tổng hợp

## Bài 1. Giá trị tương lai của khoản đầu tư

Viết chương trình cho phép:

- `pv = 500_000_000`
- `r = 0.09`
- `n = 10`

Tính giá trị tương lai của khoản đầu tư.


````python
pv = 500_000_000
r = 0.09
n = 10

fv = pv * (1 + r) ** n

print(f"Vốn ban đầu: {pv:,.0f} đồng")
print(f"Lãi suất: {r:.1%}/năm")
print(f"Số năm: {n}")
print(f"Giá trị tương lai: {fv:,.0f} đồng")
````

## Bài 2. Xếp loại cơ hội đầu tư

Cho danh sách lợi suất dự kiến:

```python
returns = [0.04, 0.08, 0.15, -0.02, 0.11]
```

Với mỗi lợi suất:

- ≥ 10% → `"Attractive"`
- 0% đến dưới 10% → `"Normal"`
- < 0% → `"Loss"`

Hãy dùng vòng lặp và `if...elif...else` để phân loại.


````python
returns = [0.04, 0.08, 0.15, -0.02, 0.11]

for r in returns:
    if r >= 0.10:
        label = "Attractive"
    elif r >= 0:
        label = "Normal"
    else:
        label = "Loss"

    print(f"Return = {r:.1%} -> {label}")
````

**Nhận xét:** Các ngưỡng phân loại trong ví dụ chỉ dùng để minh họa cú pháp điều kiện, không phải khuyến nghị đầu tư thực tế.


## Bài 3. Phân tích danh mục đơn giản

Cho dictionary:


````python
portfolio = {
    "ABC": {"price": 25_000, "quantity": 1000},
    "XYZ": {"price": 40_000, "quantity": 500},
    "DEF": {"price": 15_000, "quantity": 2000}
}

total_value = 0
largest_symbol = None
largest_value = -1

for symbol, info in portfolio.items():
    value = info["price"] * info["quantity"]
    total_value += value

    print(f"{symbol}: {value:,.0f} đồng")

    if value > largest_value:
        largest_value = value
        largest_symbol = symbol

print("-" * 40)
print(f"Tổng giá trị danh mục: {total_value:,.0f} đồng")
print(f"Cổ phiếu có giá trị đầu tư lớn nhất: {largest_symbol}")
print(f"Giá trị: {largest_value:,.0f} đồng")
````

**Nhận xét:** Giá trị vị thế bằng `price × quantity`. Tổng danh mục là tổng giá trị của các vị thế riêng lẻ.


## Bài 4. Lưu kết quả ra file

Sau khi tính tổng giá trị danh mục ở Bài 3:

1. Ghi kết quả vào `portfolio_report.txt`
2. Đọc lại file
3. Hiển thị nội dung


````python
# Tính lại giá trị danh mục
portfolio = {
    "ABC": {"price": 25_000, "quantity": 1000},
    "XYZ": {"price": 40_000, "quantity": 500},
    "DEF": {"price": 15_000, "quantity": 2000}
}

total_value = 0
lines = []

for symbol, info in portfolio.items():
    value = info["price"] * info["quantity"]
    total_value += value
    lines.append(f"{symbol}: {value:,.0f} đồng")

lines.append(f"Tổng giá trị danh mục: {total_value:,.0f} đồng")

with open("portfolio_report.txt", "w", encoding="utf-8") as file:
    file.write("\n".join(lines))

with open("portfolio_report.txt", "r", encoding="utf-8") as file:
    report = file.read()

print(report)
````
