# 2. NumPy – Numerical Python

NumPy cung cấp cấu trúc dữ liệu chính là **ndarray** (N-dimensional array).

Khác với `list` thông thường của Python, NumPy:

- tối ưu cho tính toán số học,
- hỗ trợ phép toán vector hóa,
- thao tác nhanh trên lượng lớn dữ liệu số,

## 2.1. So sánh Python list và NumPy array

Điểm khác biệt rất dễ thấy là toán tử `*`.


````python
# Python list
v_list = list(range(1, 6))
print("Python list:", v_list)

print("2 * list:")
print(2 * v_list)
````

Với `list`, phép nhân `2 * v_list` **không nhân từng phần tử với 2**.  
Nó lặp lại danh sách hai lần.


````python
# NumPy array
v = np.arange(1, 6)

print("NumPy array:", v)
print("2 * array :", 2 * v)
````

Với NumPy, phép toán được thực hiện **theo từng phần tử**:

$$
[1,2,3,4,5] \times 2 = [2,4,6,8,10]
$$

Đây gọi là **vectorized computation**.

# 6. Phép toán vector hóa

Hai mảng cùng kích thước có thể thực hiện phép toán theo từng phần tử.


````python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("a =", a)
print("b =", b)
````

````python
print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)
````

Trong đó:

```python
a * b
```

là **nhân theo từng phần tử**, không phải phép nhân ma trận.


## 6.1. Ví dụ tài chính

Giả sử ta có giá cổ phiếu và số lượng đang nắm giữ.


````python
prices = np.array([50_000, 80_000, 120_000])
quantities = np.array([1000, 500, 200])

market_values = prices * quantities

print("Giá trị từng vị thế:")
print(market_values)

print("Tổng giá trị danh mục:",
      f"{market_values.sum():,.0f} đồng")
````

# 8. Slicing mảng

Slicing giúp lấy một phần của mảng.

Cú pháp tổng quát:

```python
array[start:stop]
```

Giá trị tại `stop` không được lấy.


````python
print(a)
````

````python
# Lấy hàng thứ 3
print(a[2])
````

````python
# Lấy hàng thứ 3 nhưng vẫn giữ dạng ma trận 2 chiều
print(a[2:3, :])
````

````python
# Lấy hai cột đầu tiên
print(a[:, :2])
````

````python
# Hàng thứ 2, hai cột đầu
print(a[1, :2])
````

### Ghi nhớ

- `:` nghĩa là lấy tất cả.
- `a[2]` trả về mảng 1 chiều.
- `a[2:3, :]` vẫn giữ mảng 2 chiều.


# 9. Khái niệm `axis`

Trong mảng 2 chiều:

- `axis=0`: thao tác theo chiều **dọc**, kết quả cho từng cột.
- `axis=1`: thao tác theo chiều **ngang**, kết quả cho từng hàng.


````python
x = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(x)
````

````python
print("Tổng toàn bộ:", x.sum())
print("Tổng theo cột (axis=0):", x.sum(axis=0))
print("Tổng theo hàng (axis=1):", x.sum(axis=1))
````

### Ví dụ kinh doanh

Nếu:

- mỗi hàng = một cửa hàng,
- mỗi cột = doanh thu của một tháng,

thì:

- `sum(axis=0)` → tổng doanh thu theo tháng,
- `sum(axis=1)` → tổng doanh thu theo cửa hàng.

# 18. `groupby()` – tổng hợp dữ liệu theo nhóm

`groupby()` là một trong những thao tác quan trọng nhất trong phân tích dữ liệu kinh doanh.

Ví dụ: tổng doanh thu theo khu vực.


````python
transactions = pd.DataFrame({
    "Region": ["South", "North", "South", "Central", "North", "South"],
    "Revenue": [120, 90, 150, 70, 110, 180]
})

transactions
````

````python
revenue_by_region = transactions.groupby("Region")["Revenue"].sum()

revenue_by_region
````

Ta có thể hiểu `groupby()` theo ba bước:

1. **Split** – chia dữ liệu theo nhóm.
2. **Apply** – áp dụng phép tính.
3. **Combine** – ghép kết quả lại.

Đây cũng là tư tưởng rất quan trọng khi chuyển sang PySpark.

## 19.1. Nạp dữ liệu

Để notebook chạy ổn định không phụ thuộc Internet, ta sử dụng bản Iris tích hợp trong `scikit-learn`.


````python
from sklearn.datasets import load_iris

iris_bunch = load_iris(as_frame=True)
iris_df = iris_bunch.frame.copy()

# Đổi tên target từ số sang tên loài
iris_df["species"] = iris_df["target"].map(
    dict(enumerate(iris_bunch.target_names))
)

# Bỏ cột target số để dễ quan sát
iris_df = iris_df.drop(columns="target")

iris_df.head(10)
````

# 21. Trực quan hóa dữ liệu

Seaborn có thể tô màu điểm theo biến phân loại `species`.

# 22. Ma trận tương quan

Tương quan đo mức độ liên hệ tuyến tính giữa hai biến.

Giá trị:

\[
-1 \leq r \leq 1
\]

- gần 1: tương quan dương mạnh,
- gần -1: tương quan âm mạnh,
- gần 0: quan hệ tuyến tính yếu.


````python
corr = iris_df[numeric_cols].corr()

corr
````

````python
plt.figure(figsize=(6, 5))
plt.imshow(corr, aspect="auto")
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=45, ha="right")
plt.yticks(range(len(corr.index)), corr.index)
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
````

Trong dữ liệu tài chính, tương quan thường được sử dụng khi:

- phân tích mối liên hệ giữa các cổ phiếu,
- đa dạng hóa danh mục,
- phân tích các yếu tố kinh tế.

## 23.1. Ví dụ dữ liệu đầu tư đơn giản


````python
investment_df = pd.DataFrame({
    "Stock": ["AAA", "BBB", "CCC", "DDD", "EEE"],
    "Return": [0.12, 0.08, -0.03, 0.15, 0.06],
    "Volatility": [0.20, 0.15, 0.30, 0.25, 0.12],
    "Sector": ["Bank", "Bank", "Tech", "Tech", "Retail"]
})

investment_df
````

````python
investment_df.describe()
````

````python
investment_df.groupby("Sector")[["Return", "Volatility"]].mean()
````

````python
plt.figure(figsize=(7, 5))

for sector, group in investment_df.groupby("Sector"):
    plt.scatter(
        group["Volatility"],
        group["Return"],
        label=sector,
        s=80
    )

plt.axhline(0, linewidth=1)
plt.xlabel("Volatility")
plt.ylabel("Return")
plt.title("Risk - Return")
plt.legend()
plt.show()
````

## Bài 2. pandas

Cho dữ liệu:


````python
df_sales = pd.DataFrame({
    "Product": ["A", "B", "C", "D"],
    "Revenue": [500, 700, 450, 900],
    "Cost": [300, 450, 320, 600]
})

df_sales
````

Yêu cầu:

1. Tạo cột `Profit`.
2. Tạo cột `ProfitMargin`.
3. Tìm sản phẩm có lợi nhuận cao nhất.


````python
df_sales["Profit"] = df_sales["Revenue"] - df_sales["Cost"]
df_sales["ProfitMargin"] = df_sales["Profit"] / df_sales["Revenue"]

df_sales
````

````python
best_product = df_sales.loc[df_sales["Profit"].idxmax()]

print("Sản phẩm lợi nhuận cao nhất:")
print(best_product)
````
